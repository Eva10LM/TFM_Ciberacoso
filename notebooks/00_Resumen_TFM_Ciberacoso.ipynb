# Detección y clasificación automática de ciberacoso en redes sociales mediante Machine Learning

## Trabajo Fin de Máster

**Máster en Data Science e Inteligencia Artificial – Nodd3r**

**Autora:** Eva Laguna Martín

Este proyecto estudia la posibilidad de utilizar técnicas de Procesamiento del Lenguaje Natural (NLP) y Machine Learning clásico para analizar y clasificar automáticamente mensajes publicados en redes sociales según diferentes tipos de ciberacoso.

El trabajo incluye el análisis y limpieza de los datos, el estudio exploratorio del texto, la comparación y optimización de diferentes modelos de clasificación y el desarrollo de una aplicación web con Streamlit que permite realizar predicciones a partir de nuevos mensajes.

## Abstract

El ciberacoso es un problema especialmente relevante en el uso de redes sociales y puede presentarse de formas diferentes según el contenido de los mensajes.

El objetivo de este proyecto es estudiar si técnicas de Procesamiento del Lenguaje Natural y Machine Learning clásico permiten identificar patrones textuales asociados a distintos tipos de ciberacoso.

Se ha trabajado con un dataset de 47.692 mensajes clasificados en seis categorías. Después del análisis de calidad y la eliminación de duplicados y conflictos entre etiquetas, se utilizaron 44.378 mensajes.

Los textos se representaron mediante TF-IDF y se compararon Regresión Logística, Multinomial Naive Bayes y Linear SVM. Tras los experimentos y la optimización, se seleccionó una Regresión Logística con unigramas y bigramas.

El modelo obtuvo una Accuracy de 0,8767 y un Macro F1 de 0,8653 sobre el conjunto de test.

Finalmente, el modelo se integró en una aplicación desarrollada con Streamlit para permitir la clasificación de nuevos mensajes.

## 1. Introducción y planteamiento del problema

El uso de redes sociales forma parte de la vida cotidiana de muchas personas y tiene una presencia especialmente importante entre adolescentes y jóvenes. Junto con las posibilidades de comunicación que ofrecen estas plataformas también pueden aparecer comportamientos negativos, entre ellos el ciberacoso.

El gran volumen de mensajes que se genera en redes sociales hace difícil revisar manualmente todo el contenido. Por este motivo, las técnicas de Procesamiento del Lenguaje Natural y Machine Learning pueden ser útiles como herramientas de apoyo para analizar grandes cantidades de texto e identificar patrones relacionados con diferentes tipos de ciberacoso.

A partir de este problema se plantea la siguiente pregunta:

**¿Es posible utilizar técnicas de Procesamiento del Lenguaje Natural y Machine Learning clásico para identificar y clasificar automáticamente diferentes tipos de contenido relacionado con el ciberacoso a partir del texto de mensajes publicados en redes sociales?**

El proyecto se plantea como un problema de clasificación multiclase, en el que cada mensaje debe ser clasificado en una de las seis categorías disponibles en el conjunto de datos.

## 2. Objetivos

### Objetivo general

Desarrollar un modelo de Machine Learning capaz de analizar el contenido textual de mensajes publicados en redes sociales y clasificarlos automáticamente en diferentes categorías relacionadas con el ciberacoso.

- Analizar la calidad y estructura del conjunto de datos utilizado.
- Detectar y tratar posibles problemas como valores nulos, duplicados o textos con etiquetas contradictorias.
- Realizar un análisis exploratorio de los mensajes y de las diferentes categorías.
- Preparar y representar los textos mediante técnicas de Procesamiento del Lenguaje Natural.
- Comparar diferentes algoritmos de Machine Learning para clasificación multiclase.
- Optimizar el modelo que presente mejores resultados.
- Evaluar el modelo final utilizando diferentes métricas y analizar sus principales errores.
- Entrenar y guardar un Pipeline final que permita realizar predicciones sobre nuevos mensajes.

## 3. Dataset

Para el desarrollo del proyecto se ha utilizado el dataset **Cyberbullying Classification**, disponible públicamente en Kaggle.

El conjunto de datos original contiene **47.692 registros** y dos variables:

- `tweet_text`: contiene el texto del mensaje publicado.
- `cyberbullying_type`: contiene la categoría asignada al mensaje.

La variable objetivo presenta seis categorías:

- `age`: ciberacoso relacionado con la edad.
- `ethnicity`: ciberacoso relacionado con la etnia.
- `gender`: ciberacoso relacionado con el género.
- `religion`: ciberacoso relacionado con la religión.
- `other_cyberbullying`: otros tipos de ciberacoso.
- `not_cyberbullying`: mensajes clasificados como no ciberacoso.

Antes de utilizar los datos para el modelado se realizó un análisis de calidad para comprobar la existencia de valores nulos, duplicados y posibles inconsistencias en las etiquetas.

## 4. Calidad y limpieza de los datos

El análisis inicial mostró que el dataset no contenía valores nulos, aunque sí se detectaron diferentes problemas relacionados con los textos duplicados y las etiquetas.

Los principales resultados fueron:

- **0 valores nulos.**
- **36 filas completamente duplicadas.**
- **1.675 textos duplicados.**
- **1.639 textos asociados a más de una etiqueta.**

Este último punto era especialmente importante, ya que un mismo texto podía aparecer asociado a categorías diferentes. La mayor parte de estos conflictos se encontraba entre las categorías `not_cyberbullying` y `other_cyberbullying`.

Para evitar introducir información contradictoria durante el entrenamiento, se decidió eliminar los registros correspondientes a textos que presentaban etiquetas diferentes. Posteriormente se eliminaron también los textos duplicados restantes.

Después de aplicar este proceso de limpieza, el conjunto de datos quedó formado por **44.378 mensajes**, lo que supone la eliminación de 3.314 registros, aproximadamente un **6,95 %** del dataset original.

Tras la limpieza se volvió a comprobar la calidad de los datos y el conjunto final no presentaba valores nulos, textos duplicados ni conflictos entre etiquetas.

La distribución de las seis categorías quedó relativamente equilibrada, con porcentajes aproximados entre el 14 % y el 18 %. Por este motivo no se consideró necesario aplicar técnicas adicionales de balanceo de clases.

## 5. Análisis exploratorio de los datos

Una vez realizada la limpieza se llevó a cabo el análisis exploratorio para conocer mejor las características de los mensajes y comprobar si existían diferencias entre las seis categorías.

En primer lugar se analizó la distribución de las clases. Después de la limpieza todas las categorías quedaron representadas con porcentajes aproximados entre el 14 % y el 18 %, por lo que el conjunto final puede considerarse razonablemente equilibrado.

También se estudió la longitud de los mensajes. Se observaron diferencias entre categorías: los mensajes relacionados con religión y edad presentaban una longitud media mayor, mientras que `not_cyberbullying` y `other_cyberbullying` contenían, en general, mensajes más cortos.

Las longitudes medias aproximadas fueron:

| Categoría | Media de palabras |
|---|---:|
| Religión | 33,43 |
| Edad | 32,37 |
| Etnia | 24,89 |
| Género | 23,83 |
| Otros tipos de ciberacoso | 14,15 |
| No ciberacoso | 13,55 |

Además, se analizaron las palabras y bigramas más frecuentes de cada categoría. Este análisis permitió observar que algunas clases presentan un vocabulario bastante específico, mientras que `not_cyberbullying` y `other_cyberbullying` son más heterogéneas.

Estas diferencias sugieren que el contenido textual puede aportar información útil para distinguir entre las categorías, aunque la frecuencia de una palabra por sí sola no significa necesariamente que sea importante para realizar la clasificación.

## 6. Representación del texto con TF-IDF

Los algoritmos de Machine Learning utilizados en el proyecto no trabajan directamente con texto, por lo que fue necesario transformar los mensajes en una representación numérica.

Para ello se utilizó **TF-IDF (Term Frequency-Inverse Document Frequency)**. Esta técnica asigna un peso a las palabras teniendo en cuenta tanto su presencia en un documento como su frecuencia en el conjunto completo de documentos.

De forma simplificada, TF-IDF permite dar mayor importancia a términos que pueden aportar información para diferenciar unos mensajes de otros y reducir el peso de palabras que aparecen de forma muy generalizada.

Durante el EDA se realizó una primera exploración mediante TF-IDF para estudiar el vocabulario del conjunto de datos. Posteriormente, durante el modelado, el vectorizador se integró dentro de un `Pipeline` y se ajustó únicamente con los datos correspondientes al entrenamiento.

Esta separación es importante para evitar **data leakage o fuga de información**, es decir, que información procedente del conjunto de test influya de alguna manera durante el entrenamiento del modelo.

## 7. Modelado y experimentos

Para entrenar y evaluar los modelos se dividió el dataset limpio en dos conjuntos:

- **80 % para entrenamiento:** 35.502 mensajes.
- **20 % para test:** 8.876 mensajes.

La división se realizó de forma estratificada utilizando `random_state=42`, de manera que las seis categorías mantuvieran una distribución similar en ambos conjuntos y los experimentos pudieran reproducirse.

En los primeros experimentos se utilizó TF-IDF con unigramas, eliminación de stopwords en inglés, `min_df=5` y un máximo de 20.000 características.

Se compararon tres algoritmos de Machine Learning clásico:

- Regresión Logística.
- Multinomial Naive Bayes.
- Linear SVM (`LinearSVC`).

Los resultados obtenidos fueron:

| Modelo | Accuracy | Macro F1 |
|---|---:|---:|
| Regresión Logística | 0,8748 | 0,8636 |
| Multinomial Naive Bayes | 0,8060 | 0,7784 |
| LinearSVC | 0,8707 | 0,8578 |

La Regresión Logística obtuvo el mejor resultado inicial, aunque LinearSVC presentó un rendimiento muy próximo.

## 8. Incorporación de bigramas y selección del modelo

Después de los modelos iniciales se realizó un nuevo experimento incorporando **unigramas y bigramas** mediante `ngram_range=(1,2)` y aumentando el máximo de características a 30.000.

Los resultados fueron:

| Modelo | Accuracy | Macro F1 |
|---|---:|---:|
| Regresión Logística | 0,8754 | 0,8641 |
| LinearSVC | 0,8753 | 0,8628 |

La incorporación de bigramas produjo una mejora pequeña, pero permitió tener en cuenta combinaciones de dos palabras consecutivas y no únicamente palabras individuales.

Los resultados de Regresión Logística y LinearSVC fueron muy similares. Finalmente se seleccionó **Regresión Logística** porque obtuvo el mejor Macro F1 de los modelos comparados y ofrecía una base adecuada para continuar con la optimización.

Multinomial Naive Bayes no se incluyó en este segundo experimento porque su rendimiento inicial había sido claramente inferior al de los otros dos modelos.

## 9. Optimización del modelo

Una vez seleccionada la Regresión Logística se realizó una optimización de hiperparámetros mediante `GridSearchCV`.

La búsqueda se realizó únicamente sobre los datos de entrenamiento y utilizando validación cruzada de 3 particiones (`cv=3`). Como métrica de optimización se utilizó **Macro F1**.

Se probaron las siguientes configuraciones:

- Valores de `C`: 0,5, 1 y 2.
- TF-IDF únicamente con unigramas `(1,1)`.
- TF-IDF con unigramas y bigramas `(1,2)`.

La mejor configuración obtenida fue:

- `C = 2`.
- `ngram_range = (1,2)`.
- `min_df = 5`.
- `max_features = 30000`.

El mejor Macro F1 medio obtenido durante la validación cruzada fue aproximadamente **0,8575**.

Esta configuración se utilizó posteriormente para construir y evaluar el modelo final.

## 10. Evaluación del modelo final

La configuración seleccionada se evaluó finalmente sobre el conjunto de test, que se había mantenido separado durante el entrenamiento y la optimización.

Los resultados globales fueron:

- **Accuracy: 0,8767**
- **Macro F1: 0,8653**
- **Weighted F1: 0,8771**

El análisis por categoría mostró diferencias importantes:

| Categoría | F1-score |
|---|---:|
| Edad (`age`) | 0,9644 |
| Etnia (`ethnicity`) | 0,9744 |
| Género (`gender`) | 0,8943 |
| Religión (`religion`) | 0,9563 |
| Otros tipos de ciberacoso (`other_cyberbullying`) | 0,7317 |
| No ciberacoso (`not_cyberbullying`) | 0,6706 |

Las categorías relacionadas con edad, etnia y religión presentan los mejores resultados. Por el contrario, las mayores dificultades aparecen en `not_cyberbullying` y `other_cyberbullying`.

La matriz de confusión confirmó este comportamiento. Los errores más frecuentes se producen precisamente entre estas dos categorías:

- 268 mensajes de `not_cyberbullying` fueron clasificados como `other_cyberbullying`.
- 213 mensajes de `other_cyberbullying` fueron clasificados como `not_cyberbullying`.

Este resultado es coherente con lo observado durante el análisis exploratorio, ya que estas dos categorías contienen mensajes más generales y heterogéneos y presentan un vocabulario menos específico que otras clases.

## 11. Entrenamiento y guardado del modelo definitivo

Una vez finalizada la evaluación y seleccionada la configuración definitiva, se volvió a entrenar el Pipeline utilizando los **44.378 mensajes disponibles**.

El Pipeline final está formado por:

1. Un vectorizador TF-IDF con unigramas y bigramas.
2. Una Regresión Logística con `C=2`.

El modelo entrenado se guardó mediante `joblib` en el archivo:

`modelo_ciberacoso.pkl`

Este archivo es el que posteriormente utiliza la aplicación Streamlit para realizar predicciones sobre nuevos mensajes.

Es importante diferenciar este entrenamiento final de la evaluación anterior. Las métricas presentadas corresponden al modelo evaluado sobre el conjunto de test antes de volver a entrenar el Pipeline con todos los datos disponibles.

## 12. Aplicación y despliegue con Streamlit

Como parte final del proyecto se desarrolló una aplicación web sencilla utilizando Streamlit con el objetivo de poder probar el modelo de forma interactiva.

La aplicación permite escribir un mensaje y obtener la categoría predicha por el modelo. Para realizar la predicción se carga el archivo `modelo_ciberacoso.pkl`, que contiene el Pipeline completo con la transformación TF-IDF y la Regresión Logística.

Las etiquetas originales del dataset se muestran en la aplicación mediante descripciones en español para facilitar la interpretación del resultado.

La aplicación se desplegó públicamente mediante Streamlit Community Cloud utilizando el repositorio del proyecto alojado en GitHub.

**Aplicación:**  
https://eva10lm-tfm-ciberacoso-app-flexfp.streamlit.app

La aplicación debe considerarse un prototipo para analizar patrones textuales. La predicción realizada por el modelo no sustituye una valoración profesional ni permite determinar por sí sola la existencia de una situación real de ciberacoso.

## 13. Conclusiones

Los resultados obtenidos permiten responder de forma positiva a la pregunta planteada al inicio del proyecto: las técnicas de Procesamiento del Lenguaje Natural y Machine Learning clásico permiten identificar patrones útiles para clasificar automáticamente mensajes en las categorías de ciberacoso disponibles en el dataset.

Entre los modelos comparados, la Regresión Logística combinada con TF-IDF obtuvo el mejor rendimiento global. Después de la optimización, el modelo final alcanzó una Accuracy de 0,8767 y un Macro F1 de 0,8653.

Sin embargo, el análisis por categorías ha sido especialmente importante. Aunque algunas clases como edad, etnia o religión presentan valores F1 superiores a 0,95, el rendimiento disminuye en `not_cyberbullying` y `other_cyberbullying`.

Esto demuestra que una métrica global por sí sola no es suficiente para valorar el comportamiento del modelo y que es necesario analizar también los resultados por clase y los errores de clasificación.

Otro aspecto importante del proyecto ha sido el análisis previo de la calidad de los datos. La detección de textos duplicados y, especialmente, de mensajes asociados a etiquetas contradictorias permitió preparar un conjunto de datos más consistente antes de comenzar el entrenamiento.

Finalmente, el desarrollo de la aplicación Streamlit ha permitido completar el proceso desde el análisis inicial de los datos hasta el despliegue de un modelo capaz de recibir nuevos textos y generar una predicción.

## 14. Limitaciones y posibles mejoras

El proyecto presenta varias limitaciones que deben tenerse en cuenta al interpretar los resultados.

En primer lugar, el rendimiento es claramente inferior en las categorías `not_cyberbullying` y `other_cyberbullying`. Ambas contienen mensajes más heterogéneos y son también las categorías entre las que se produce un mayor número de errores.

El dataset procede de Twitter y contiene principalmente mensajes en inglés. Por tanto, los resultados obtenidos no permiten afirmar que el modelo tenga el mismo rendimiento con mensajes en español, otros idiomas o textos procedentes de otras redes sociales.

Durante el análisis exploratorio también se observaron términos y expresiones muy específicos del propio conjunto de datos. Esto puede limitar la capacidad de generalización cuando el modelo se utiliza con mensajes procedentes de contextos diferentes.

Como posibles líneas de mejora se plantean:

- Incorporar datos procedentes de otras redes sociales.
- Ampliar el dataset con mensajes en español y otros idiomas.
- Analizar con mayor profundidad la separación entre `not_cyberbullying` y `other_cyberbullying`.
- Evaluar el modelo utilizando conjuntos de datos externos.
- Probar otras técnicas de representación y procesamiento del lenguaje.
- Estudiar métodos adicionales que permitan mejorar la capacidad de generalización del modelo.

En un contexto real, una herramienta de este tipo debería utilizarse como apoyo al análisis y no como un sistema automático para tomar decisiones sobre personas o situaciones concretas.
